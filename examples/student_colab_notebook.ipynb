## Step 2 — Connect to the classroom AI
Fill in the password your teacher gives you (replace `ASK-YOUR-TEACHER`).

Notice there is **no `sk-...` key anywhere** — that's the point! 🎉

```python
import expressai

PROXY_URL = "https://franklin-ai-proxy-c953cd639dde.herokuapp.com/v1"   # your class proxy
CLASSROOM_PASSWORD = "ASK-YOUR-TEACHER"   # <-- your teacher will give you this password
MY_NAME = "your-first-name"                # <-- pick a nickname

bot = expressai.create_chatbot(
    system_prompt="You are a friendly and encouraging study buddy for a 13-year-old.",
    base_url=PROXY_URL,
    classroom_token=CLASSROOM_PASSWORD,
    student_id=MY_NAME,
)

print("Connected! Your bot is ready.")
print(bot)
```

## Step 3 — Say hello
Ask your bot anything. You can call `bot("...")` or `bot.ask("...")` — they do the same thing.

```python
answer = bot.ask("Hi! Can you help me understand what a variable is in coding?")
print(answer)
```

## Step 4 — Give your bot a personality
The `system_prompt` is where you describe *how* your bot should act. Try changing it!

```python
pirate_bot = expressai.create_chatbot(
    system_prompt="You are a jolly pirate captain who explains coding using sea adventures.",
    base_url=PROXY_URL,
    classroom_token=CLASSROOM_PASSWORD,
    student_id=MY_NAME,
)

print(pirate_bot.ask("What is a loop in programming?"))
```

## Step 5 — Have a real conversation (memory!)
Your bot remembers what you said earlier in the chat. This loop keeps talking until you type **bye**.

Run the cell, then type in the box that appears.

```python
print("Chat with your bot! Type 'bye' to stop.\n")

while True:
    my_message = input("You: ")
    if my_message.lower() in ("bye", "quit", "exit"):
        print("Bot: Goodbye! 👋")
        break
    reply = bot.ask(my_message)
    print("Bot:", reply)
    print()
```

## Step 6 — Start fresh (optional)
Want your bot to forget the conversation and start over? Run this.

```python
bot.reset()
print("Memory cleared. Your bot forgot the conversation.")
```
