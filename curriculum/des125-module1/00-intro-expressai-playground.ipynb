## Part 2 — Connect to the class AI (no API key!)
Edit the **two lines** with the arrows, then run the cell.
- `CLASSROOM_PASSWORD` — the password your teacher gives you.
- `MY_NAME` — any nickname (used so everyone gets a fair share of the AI).

Notice there is **no `sk-...` key anywhere.** That's the point. 🎉

```python
import expressai

PROXY_URL = "https://franklin-ai-proxy-c953cd639dde.herokuapp.com/v1"
CLASSROOM_PASSWORD = "ASK-YOUR-TEACHER"   # <-- type the class password here
MY_NAME = "your-nickname"                 # <-- pick a nickname

bot = expressai.create_chatbot(
    system_prompt="You are a friendly, encouraging study buddy for a 9th grader.",
    base_url=PROXY_URL,
    classroom_token=CLASSROOM_PASSWORD,
    student_id=MY_NAME,
)

print("Connected! Your bot is ready.")
```

## Part 3 — Say something to your bot
`bot.ask("...")` sends a message and gives you the answer back. Run it, then change the question and run again.

```python
print(bot.ask("Hi! Can you help me understand what a variable is?"))
```

## Part 4 — The `system_prompt` is your bot's personality
When you made the bot, you gave it a `system_prompt` — that's the instruction that decides *how* it behaves. Change it and you change the bot. Run this pirate to see:

```python
pirate_bot = expressai.create_chatbot(
    system_prompt="You are a jolly pirate captain who explains things with sea adventures.",
    base_url=PROXY_URL,
    classroom_token=CLASSROOM_PASSWORD,
    student_id=MY_NAME,
)

print(pirate_bot.ask("What is a loop in programming?"))
```

### Have a real conversation
Run the cell below, then type in the box that appears. Type **bye** to stop.

```python
print("Chat with your bot! Type 'bye' to stop.\n")

while True:
    my_message = input("You: ")
    if my_message.lower() in ("bye", "quit", "exit"):
        print("Bot: Goodbye! 👋")
        break
    print("Bot:", bot.ask(my_message))
    print()
```

Want your bot to forget everything and start fresh? Run this:

```python
bot.reset()
print("Memory cleared — your bot forgot the conversation.")
```
