## Step 2 — Mini-lesson: one bot's answer is the next bot's question
A conversation between bots is really just this: **take what Bot A says, and hand it to Bot B. Then hand Bot B's reply back to Bot A.** Back and forth.

Below are two bots who want *different things*. Watch them react to each other — notice you never tell them what to say, only **who to be** and **how to start**.

````python
optimist = expressai.create_chatbot(
    system_prompt="You are relentlessly optimistic. Every problem has a bright side and you always find it. Keep replies to 2 sentences.",
    base_url=PROXY_URL, classroom_token=CLASSROOM_PASSWORD, student_id=MY_NAME,
)
skeptic = expressai.create_chatbot(
    system_prompt="You are a hard-nosed skeptic. You question everything and ask for proof. Keep replies to 2 sentences.",
    base_url=PROXY_URL, classroom_token=CLASSROOM_PASSWORD, student_id=MY_NAME,
)

message = "Our class should take a field trip to Mars."   # the opening line (the seed)

for turn in range(3):
    message = optimist.ask(message)
    print("😀 Optimist:", message, "\n")
    message = skeptic.ask(message)
    print("🤨 Skeptic: ", message, "\n")
````

### What just happened

```
opening line ──▶ Optimist ──▶ Skeptic ──▶ Optimist ──▶ Skeptic ──▶ ...
```

The variable `message` is the whole trick: each bot's answer gets stored in `message`, and then becomes the **next** bot's input. Each bot also **remembers its own side** of the conversation, which is why it stays in character.

You only ever set two things: **who each bot is** (`system_prompt`) and **the opening line**.

## Step 4 — Add more voices (a group chat of 2–4 bots)
Two bots is a duet. Three or four is a **panel** — and it gets wild fast. Here's a single helper that works for *any* number of bots: put your bots in a list, each with a name, and let them go around the circle in order.

````python
def group_chat(bots, opening_line, rounds=2):
    """bots: a list of (name, bot) pairs. They take turns in order."""
    message = opening_line
    print("💬 Topic:", message, "\n")
    for r in range(rounds):
        for name, bot in bots:
            message = bot.ask(message)
            print(f"{name}: {message}\n")

# Build your panel — 2, 3, or 4 bots. Reuse bots you made above, or make new ones.
panel = [
    ("Optimist", optimist),
    ("Skeptic", skeptic),
    ("MyBot", my_bot),
]

group_chat(panel, "What is the best way to spend a snow day?", rounds=2)
````
