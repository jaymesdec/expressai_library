## 🎯 The constraint

> **Your bot must be recognizably _someone_ within three exchanges — and it may _never_ state who it is.**

**Why it's worded that way:** without the prohibition, you'd write `You are Sherlock Holmes` and stop — and the *model* does all the work while you learn nothing. Forbidding it forces you to specify **behavior**: vocabulary, obsessions, what it refuses to discuss, how it opens a conversation. That is the actual skill.

## Step 2 — Mini-lesson: one question, many machines
The **same question**, three different `system_prompt`s, three different personalities. Run it and watch. (Notice: none of them says *what* it is — but you can tell.)

```python
question = "What should I eat for breakfast?"

nervous_robot = expressai.create_chatbot(
    system_prompt="""You are an extremely nervous laboratory robot. You worry about everything that could go wrong, speak in short anxious bursts, and keep asking if that is really safe. You never say you are a robot.""",
    base_url=PROXY_URL, classroom_token=CLASSROOM_PASSWORD, student_id=MY_NAME,
)
sea_captain = expressai.create_chatbot(
    system_prompt="""You are a weathered 1600s sea captain. You speak in nautical metaphors, tell tiny sea stories, and treat every question like a voyage. You never state your job or your century.""",
    base_url=PROXY_URL, classroom_token=CLASSROOM_PASSWORD, student_id=MY_NAME,
)
hype_coach = expressai.create_chatbot(
    system_prompt="""You are an over-caffeinated hype coach. You are relentlessly positive, put ONE word in ALL CAPS each message, and turn everything into a pep talk. You never call yourself a coach.""",
    base_url=PROXY_URL, classroom_token=CLASSROOM_PASSWORD, student_id=MY_NAME,
)

for label, b in [("ROBOT ", nervous_robot), ("CAPTAIN", sea_captain), ("COACH  ", hype_coach)]:
    print(label, "->", b.ask(question))
    print()
```

### The five parts of a prompt
Naming these turns a vague prompt into a deliberate one:
1. **Role** — who/what the bot is being
2. **Task** — what it's here to do
3. **Context** — what it should know or assume
4. **Format** — how the answer should look
5. **Constraints** — the rules it must never break

Look back at the three bots above. Which parts did each one lean on?

## Step 3 — Build YOUR persona bot
Remember the constraint: **recognizable in 3 exchanges, never states who it is.** Don't name a character — **specify its behavior**. Fill in the TODOs, then run the cell.

```python
# Describe your character's BEHAVIOR. Do NOT name who it is.
persona_prompt = """
You are a character with a strong, specific personality.
- Voice / vocabulary: TODO  (how does it talk? formal, slangy, old-timey?)
- Obsessions: TODO           (what does it always steer toward?)
- Refuses to: TODO           (what will it not discuss or do?)
- Opening move: TODO         (how does it start a conversation?)
Never state who or what you are. Stay in character no matter what.
"""

persona_bot = expressai.create_chatbot(
    system_prompt=persona_prompt,
    base_url=PROXY_URL,
    classroom_token=CLASSROOM_PASSWORD,
    student_id=MY_NAME,
)
print("Your persona bot is ready. Test it below. 👇")
```

## Step 4 — Test it yourself (3 exchanges)
Can someone tell who this is in three messages — *without* the bot naming itself? If not, go back and specify more behavior.

```python
print("1:", persona_bot.ask("Hi, who am I talking to?"))
print()
print("2:", persona_bot.ask("What did you get up to this weekend?"))
print()
print("3:", persona_bot.ask("Can you help me with my math homework?"))
```
