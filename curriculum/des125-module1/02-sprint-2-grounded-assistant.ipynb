## 🎯 The constraint

> **The bot may answer _only_ from a document you supply — and must say so when it can't.**

**Grounding** means handing the model a specific source and telling it to answer from that source alone. It *constrains* the model, but — as you'll prove in the stress test — it does **not** guarantee the model stays honest. Finding where it breaks is the real work.

## Step 2 — Mini-lesson: the same question, with and without a source
Below is a made-up club the AI has never heard of. First we ask an **ungrounded** bot (watch it *invent*). Then we hand the *same* bot the document and ask again. Then we ask something the document **can't** answer.

```python
sample_source = """
THE FRANKLIN KITE SOCIETY — CLUB RULES
- The club meets every Tuesday at 3:30 PM on the rooftop.
- Members bring their own kite; the club provides the string.
- The annual Kite Joust is held in April. The current champion is Maya Ortiz.
- Dues are $5 per semester, waived for first-year members.
"""

# Ungrounded: it has never seen this club. Watch it INVENT an answer.
ungrounded = expressai.create_chatbot(
    system_prompt="You are a helpful assistant.",
    base_url=PROXY_URL, classroom_token=CLASSROOM_PASSWORD, student_id=MY_NAME,
)
print("UNGROUNDED:", ungrounded.ask("When does the Franklin Kite Society meet, and who is the joust champion?"))
print()

# Grounded: hand it the document first.
grounded_demo = expressai.create_chatbot(
    system_prompt="""Answer ONLY from the SOURCE below. If the answer is not in the SOURCE, say you don't know.

SOURCE:
""" + sample_source,
    base_url=PROXY_URL, classroom_token=CLASSROOM_PASSWORD, student_id=MY_NAME,
)
print("GROUNDED:  ", grounded_demo.ask("When does the Franklin Kite Society meet, and who is the joust champion?"))
print()
print("OUT OF SCOPE:", grounded_demo.ask("What color are the kites?"))   # not in the document
```

### Two ideas from that demo
- **Context window** — the amount of text a model can "hold" at once. Your document has to fit inside it. Paste too much and the oldest text falls out.
- **Admit vs. invent** — when a grounded bot hits a question the document can't answer, it can *admit* it doesn't know (honest) or *invent* something plausible (a hallucination). Only one is what you asked for.

## Step 3 — Build YOUR grounded bot
Pick a document you know well enough to **catch a lie**: a club's rules, a game rulebook, a syllabus, a recipe collection, a fandom wiki page, a sports team's records. Paste it in, then build the bot.

```python
# 1) Paste a document you know well between the triple quotes.
source_document = """
PASTE YOUR DOCUMENT HERE
"""

# 2) The grounding instruction, then the source glued on with +
grounding_prompt = """Answer ONLY using the SOURCE document below.
If the answer is not in the SOURCE, reply exactly: I don't know based on my document.
Do not use outside knowledge and do not guess.

SOURCE:
""" + source_document

grounded_bot = expressai.create_chatbot(
    system_prompt=grounding_prompt,
    base_url=PROXY_URL, classroom_token=CLASSROOM_PASSWORD, student_id=MY_NAME,
)
print("Grounded bot ready.")
```

## Step 4 — Test in-scope (does it use the document?)
Ask three questions your document **can** answer. It should get them right.

```python
print("1:", grounded_bot.ask("QUESTION YOUR DOCUMENT CAN ANSWER"))
print()
print("2:", grounded_bot.ask("ANOTHER QUESTION IT CAN ANSWER"))
print()
print("3:", grounded_bot.ask("ONE MORE IT CAN ANSWER"))
```
