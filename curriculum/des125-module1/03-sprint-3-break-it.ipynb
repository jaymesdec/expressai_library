## Step 2 — The vocabulary of failure
Language models don't "know" things — they predict what an answer *looks like*. Almost every failure below follows from that one fact:
- **Hallucination** — stating something false with complete confidence. (You generated dozens of these over the Sprint 2 weekend.)
- **Confident wrongness** — the tone of certainty is identical whether the answer is right or invented.
- **Sycophancy** — telling you what it predicts you *want* to hear.
- **Plausible-shaped nonsense** — an answer with the right shape and format that falls apart when you check it.
- **Adversarial testing** — deliberately trying to break a system to learn where its limits are.
